```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

# 1. LOAD DATA
df_air = pd.read_csv('Airbnb_Open_Data.csv', low_memory=False)
df_crime = pd.read_csv('NYPD_Complaint_Data_Current_(Year_To_Date)_20260331.csv', low_memory=False)

# 2. CLEAN AIRBNB DATA (Fixing the $ and comma issue)
# Removing '$', replacing ',' with '.', and converting to numeric
df_air['price'] = df_air['price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '.', regex=False).str.strip()
df_air['price'] = pd.to_numeric(df_air['price'], errors='coerce')

# Standardizing Borough names
valid_boroughs = ['Manhattan', 'Brooklyn', 'Queens', 'Bronx', 'Staten Island']
df_air['neighbourhood group'] = df_air['neighbourhood group'].str.title().str.strip()
df_air = df_air[df_air['neighbourhood group'].isin(valid_boroughs)].dropna(subset=['price'])

# 3. CLEAN CRIME DATA
df_crime['BORO_NM'] = df_crime['BORO_NM'].str.title().str.strip()
crime_counts = df_crime['BORO_NM'].value_counts().reset_index()
crime_counts.columns = ['neighbourhood group', 'total_crimes']

# 4. MERGE FOR COMPARISON
avg_prices = df_air.groupby('neighbourhood group')['price'].mean().reset_index()
merged_data = pd.merge(avg_prices, crime_counts, on='neighbourhood group')

# 5. VISUALIZATION: Crime vs Price
fig, ax1 = plt.subplots(figsize=(12, 6))

# Bar Chart for Crime
sns.barplot(x='neighbourhood group', y='total_crimes', data=merged_data, ax=ax1, palette='Reds_r', alpha=0.6)
ax1.set_ylabel('Total Crime Count', color='darkred', fontsize=12)

# Line Chart for Price
ax2 = ax1.twinx()
sns.lineplot(x='neighbourhood group', y='price', data=merged_data, ax=ax2, color='blue', marker='o', linewidth=3)
ax2.set_ylabel('Average Airbnb Price ($)', color='blue', fontsize=12)

plt.title('NYC Analysis: Crime Levels vs. Average Airbnb Prices')
plt.show()

# 6. HYPOTHESIS TEST (Manhattan vs Brooklyn)
manhattan = df_air[df_air['neighbourhood group'] == 'Manhattan']['price']
brooklyn = df_air[df_air['neighbourhood group'] == 'Brooklyn']['price']
p_val = stats.ttest_ind(manhattan, brooklyn, nan_policy='omit').pvalue

print(f"\nStatistical Significance (P-Value): {p_val}")
print("\nMerged Summary Table:")
print(merged_data)
```

```python
from scipy import stats

manhattan_prices = df_air[df_air['neighbourhood group'] == 'Manhattan']['price'].dropna()
brooklyn_prices = df_air[df_air['neighbourhood group'] == 'Brooklyn']['price'].dropna()

t_stat, p_val = stats.ttest_ind(manhattan_prices, brooklyn_prices, equal_var=False)

print("--- HYPOTHESIS TEST RESULTS ---")
print(f"T-statistic: {t_stat}")
print(f"P-value: {p_val}")

alpha = 0.05
if p_val < alpha:
    print("\nCONCLUSION: Since P-value < 0.05, we REJECT the null hypothesis.")
    print("There is a statistically significant difference between Manhattan and Brooklyn prices.")
else:
    print("\nCONCLUSION: Since P-value > 0.05, we FAIL TO REJECT the null hypothesis.")
    print("The price difference between the two boroughs is not statistically significant.")
```
